```python
import os
import sys
import numpy as np
import pandas as pd
import math
from scipy import signal
from scipy.integrate import odeint

import matplotlib.pyplot as plt
# import seaborn as sns

# workspace_path = os.path.dirname(os.getcwd())
# local_package_path = os.path.join(workspace_path, "DataGeneration")
# sys.path.append(local_package_path)

import ode_auto
```

## Generate the lorenz system

```python
lorenz_coeff = [[-10, 10], [28, -1, -1], [1, -8 / 3]]
lorenz_names = [
    ["x1", "x2"],
    ["x1", "x2", "x1x3"],
    ["x1x2", "x3"],
]

n = 25000
dt = 0.001
snr = 49
t = np.arange(0, float(n) * dt, dt)

out_odeint = ode_auto.solve_ode_odeint(lorenz_coeff, lorenz_names, [2, 0, 15], t)

fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(111, projection="3d")
ax.plot(out_odeint[:, 0], out_odeint[:, 1], out_odeint[:, 2], lw=0.5, color='#481b51')
ax.set_xlabel("$x_1$")
ax.set_ylabel("$x_2$")
ax.set_zlabel("$x_3$", labelpad=-2)

# Set the ticks for the x, y, and z axes
ax.set_xticks(np.arange(math.floor(min(out_odeint[:, 0])), math.ceil(max(out_odeint[:, 0])), step=7))
ax.set_yticks(np.arange(math.floor(min(out_odeint[:, 1])), math.ceil(max(out_odeint[:, 1])), step=7))
ax.set_zticks(np.arange(math.floor(min(out_odeint[:, 2])), math.ceil(max(out_odeint[:, 2])), step=7))

# Adjust the padding
plt.subplots_adjust(left=0.05, right=0.95, bottom=0.05, top=0.95)

# Save the plot to a specific directory
plt.savefig('/Users/yuzhengzhang/Desktop/Manuscripts/MS-Annual-Review-2024/presentation/img/systems-presentation/lorenz.png')
plt.show()
```
