```python
import shutil
import subprocess
from pathlib import Path

pdf_dir = Path("Plots/si-distribution-plots")
pdf_files = sorted(pdf_dir.glob("*.pdf"))

if not pdf_files:
    raise FileNotFoundError(f"No PDF files found in {pdf_dir.resolve()}")

magick = shutil.which("magick")
pdftocairo = shutil.which("pdftocairo")

if magick is None and pdftocairo is None:
    raise RuntimeError(
        "Neither 'magick' nor 'pdftocairo' is available. Install ImageMagick or Poppler to create PNG files."
    )

generated_pngs = []

for pdf_path in pdf_files:
    png_path = pdf_path.with_suffix(".png")

    if magick is not None:
        command = [
            magick,
            "-density",
            "300",
            str(pdf_path),
            "-background",
            "white",
            "-alpha",
            "remove",
            "-alpha",
            "off",
            "-quality",
            "100",
            str(png_path),
        ]
        subprocess.run(command, check=True)
    else:
        output_base = pdf_path.with_suffix("")
        command = [
            pdftocairo,
            "-png",
            "-r",
            "300",
            str(pdf_path),
            str(output_base),
        ]
        subprocess.run(command, check=True)
        generated_page = output_base.with_name(f"{output_base.name}-1.png")
        if generated_page.exists():
            generated_page.replace(png_path)

    generated_pngs.append(str(png_path))

generated_pngs
```
