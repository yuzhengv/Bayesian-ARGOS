## Legend for Bayesian-ARGOS Identified Terms Plots


```r
# Define function number and create refined legend
function_number <- 3

# Create a refined color palette for identified terms
identified_terms_colors <- colorRampPalette(c(
    "#0570b0", # Deep blue
    "#3690c0", # Medium blue
    "#74a9cf" # Light blue
))(function_number)

# Initialize legend components
legend_labels <- c("Ground Truth")
legend_colors <- c("#807dba")
legend_borders <- c("#6a51a3")

# Add equation-specific colors to legend
for (eq_idx in 1:function_number) {
    legend_labels <- c(legend_labels, paste("Equation", eq_idx))
    legend_colors <- c(legend_colors, identified_terms_colors[eq_idx])
    legend_borders <- c(
        legend_borders,
        adjustcolor(identified_terms_colors[eq_idx], alpha.f = 1)
    )
}

# Define legend properties
legend_title <- "Identified Terms in Each Equation via Bayesian-ARGOS"
legend_ncol <- min(length(legend_labels), 4) # Optimal column layout

# Save the legend plot
output_dir <- "./ResultsAnalysis/Organise-Plots/identified-terms-plots/imgs"
legend_output_path <- paste0(output_dir, "/bayesian_argos_legend.png")

png(legend_output_path, width = 12, height = 1, units = "in", res = 600)

par(mar = c(0.5, 0.5, 0.5, 0.5))
# Create a separate legend plot
plot.new()
plot.window(xlim = c(0, 1), ylim = c(0, 1))

# Create the legend with refined positioning and styling
legend(
    x = "center", # Center the legend in the plot
    legend = legend_labels,
    fill = legend_colors,
    border = legend_borders,
    horiz = FALSE,
    ncol = legend_ncol,
    cex = 1.0, # Improved text size
    bty = "n", # No border box
    title = legend_title,
    title.col = "black",
    title.cex = 1.2, # Larger title
    title.adj = 0.5, # Center title
    x.intersp = 1.2, # Better horizontal spacing
    y.intersp = 1.5, # Better vertical spacing
    xjust = 0.5,
    yjust = 0.5
)

dev.off()

print(paste("Legend plot saved to:", legend_output_path))
```
