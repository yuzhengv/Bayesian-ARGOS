```python
import os
import sys

# Add LaTeX binary path to environment
latex_path = "/Library/TeX/texbin"
if latex_path not in os.environ.get("PATH", ""):
    os.environ["PATH"] = f"{latex_path}:{os.environ.get('PATH', '')}"
    print(f"Added {latex_path} to PATH")

current_dir = os.getcwd()
project_root = os.path.abspath(os.path.join(current_dir, "../../.."))
sys.path.insert(0, project_root)

from jupyter_tikz import TexDocument
```

```python
code = r"""\documentclass[border={0mm 0mm 0mm 0mm}]{standalone}
	\usepackage{graphicx}
	\usepackage{tikz}
	\usepackage{amsfonts}
	\usepackage{amsmath}
	\usepackage{helvet}
	\renewcommand{\familydefault}{\sfdefault}
	\usetikzlibrary{quotes,angles,calc}
	\newcommand\Fontspecific{\fontsize{18}{20}\selectfont\sffamily\bfseries}
	\newcommand\FontspecificTag{\fontsize{18}{19}\selectfont\sffamily\bfseries}
	\begin{document}

\scalebox{1}{ % Use 1.5 or 2 with caution
	\begin{tikzpicture}

		% Font fallback
		\providecommand{\Fontspecific}{\fontsize{16pt}{18pt}\selectfont}

		% Define lengths
		\newlength{\rowgap}
		\setlength{\rowgap}{0.65cm}
		
		\newlength{\panelgap}
		\setlength{\panelgap}{1cm}    

		\newlength{\imgwidth}
		\setlength{\imgwidth}{\textwidth}

		\newlength{\legendwidth}
		\setlength{\legendwidth}{2\imgwidth}

		\newlength{\captionyshift}
		\setlength{\captionyshift}{0.5cm}
		
		\newlength{\plotyshift}
		\setlength{\plotyshift}{1.25cm}
		
		\newlength{\labelxshift}
		\setlength{\labelxshift}{0.65cm}    


		% Define column coordinates
		\coordinate (x_0) at (0,0);
		\coordinate (x_1) at (\imgwidth,0);

		% Define row coordinate
		\coordinate (y_0) at (0, \plotyshift + \imgwidth +\panelgap);
		\coordinate (y_1) at (0, \plotyshift + \imgwidth+ \captionyshift + \rowgap + \panelgap);
		
		% Bottom Row - Legends
		\node[anchor=north west] at ($(x_0)+(0,1.5\plotyshift)$) {\includegraphics[width=\legendwidth, trim=20pt 50pt 20pt 20pt,clip]{Plots/BA_legend_Two_Level_simplified.png}};
		% First Row
		\node[anchor=north west] at (y_0) {\includegraphics[width=\imgwidth, trim=135pt 135pt 135pt 20pt,clip]{imgs/dadras_circular_plot_n_ba.png}};
		\node[anchor=north west] at ($(y_0)+(x_1)$) {\includegraphics[width=\imgwidth, trim=135pt 135pt 135pt 20pt,clip]{imgs/rossler_circular_plot_snr_ba.png}}; 
		  
		\node[anchor=center, font=\FontspecificTag] at ($(y_0)+(0.75\labelxshift, \captionyshift)$) {a};
		\node[anchor=center, font=\Fontspecific] at ($(y_0)+(0.5\imgwidth, \captionyshift)$) {Dadras (Varying $\mathbf{n}$)};
		\node[anchor=center, font=\Fontspecific] at ($(y_0)+(1.5\imgwidth, \captionyshift)$) {R\"ossler (Varying SNR)};

		% Define the title of the plot
		\node[anchor=north west, font=\Fontspecific] at (y_1) {};
	\end{tikzpicture}
}

\end{document}"""

n_snr_document = TexDocument(code)
n_snr_document.run_latex(
    tex_program="pdflatex",
    dpi=300,
    tex_args="-interaction=nonstopmode -halt-on-error",  # Change -sc value here
    keep_temp=False,
    full_err=False,
    save_image="Plots/investigate_success_rate_decrease",
    # save_pdf="Plots/investigate_success_rate_decrease.pdf",
)
```
